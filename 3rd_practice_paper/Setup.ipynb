```python
%sql
-- 1. Create the DEV catalog (writable target) ----------
CREATE CATALOG IF NOT EXISTS global_airline
  COMMENT 'DEV catalog for the Global Airline analytics platform.';

USE CATALOG global_airline;

-- 2. Create bronze / silver / gold schemas ------------
CREATE SCHEMA IF NOT EXISTS bronze
  COMMENT 'Stores appended, un-parsed historical data maintaining raw state.';

CREATE SCHEMA IF NOT EXISTS silver
  COMMENT 'Stores cleansed, filtered, and transformed data ready for analytics.';

CREATE SCHEMA IF NOT EXISTS gold
  COMMENT 'Stores business-level aggregated data optimized for reporting and dashboards.';

-- Create the raw volume for file ingestion
CREATE VOLUME IF NOT EXISTS bronze.raw
  COMMENT 'Managed volume for staging raw incoming files.';

CREATE VOLUME IF NOT EXISTS bronze.pipeline_state
  COMMENT 'Managed volume for for managing pipeline state.';
```
